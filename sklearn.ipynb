
# Estimator
An Estimator is the base object for any algorithm in Scikit-learn that learns from data. Think of it as the "brain" or the "model" itself.

Purpose: To "learn" patterns from your dataset. This includes all machine learning models, whether they are for classification, regression, or clustering.

# Key Method: fit(X, y)

This is the learning method.

It takes the training data X (features) and, for supervised learning, the target labels y (answers).

It adjusts the internal parameters of the model to best fit the data.

# Prediction Method: predict(X)

After the estimator is fitted, you use this method to make predictions on new, unseen data X.

Example Estimators: LinearRegression(), LogisticRegression(), KMeans()

# Transformer
A Transformer is a special type of estimator that changes or preprocesses your data. It doesn't just learn from the data; it uses what it learns to create a new version of the data.

Purpose: To clean, scale, encode, or extract features. Transformers are the key components of preprocessing pipelines.

# Key Methods:

fit(X, y): The learning step. It learns the parameters from the data (e.g., it finds the mean and standard deviation of a feature).

transform(X): The changing step. It uses the already learned parameters to change the data X (e.g., it scales the feature using its mean and standard deviation).

fit_transform(X, y): A convenient and efficient method that performs both the fit and transform steps at the same time. You almost always use this on your training data.

Example Transformers: StandardScaler(), MinMaxScaler(), OneHotEncoder(), PCA()
