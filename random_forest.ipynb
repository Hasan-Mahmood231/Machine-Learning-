# Random Forest :

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('tested.csv')
```

```python
df.head()
```

- transform the sex col to 0,1

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df['Sex'] = le.fit_transform(df['Sex'])
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
from sklearn.preprocessing import Binarizer

# 1. Initialize with your "threshold" (45)
# Note: Binarizer marks everything ABOVE the threshold as 1.
# Since you want BELOW 45 to be 1, we use a small trick.
bn = Binarizer(threshold=44) 

# 2. Apply it (Needs a 2D array, so we use [[]])
df['Age'] = bn.fit_transform(df[['Age']])
```

```python
x = df[['Age','Sex']]
y = df['Survived']
```

- now split the data and make decision.

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
# Train-Test Split (80% train, 20% test)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)   
```

```python
x.shape
```

```python
df1 = x.sample(frac=0.2, random_state=42).reset_index(drop=True)
df2 = x.sample(frac=0.2, random_state=42).reset_index(drop=True)
df3 = x.sample(frac=0.2, random_state=42).reset_index(drop=True)
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
```
