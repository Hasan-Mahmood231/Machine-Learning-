```python
from sklearn.linear_model import LogisticRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
# X = Age, y = Buy (1) or Not Buy (0)
x = np.array([[18], [20], [25], [35], [40], [50], [60]])
y = np.array([0, 0, 0, 1, 1, 1, 1]) 
```

```python
model = LogisticRegression()
model.fit(x,y)
age_to_test = [[32]]
prediction = model.predict(age_to_test)
probability = model.predict_proba(age_to_test)
```

```python
plt.scatter(x, y, color='red', label='Actual Data (0=Fail, 1=Pass)')
```

```python
print(f"customer age :{age_to_test[0][0]}")
print(f"well they buy {'yes' if prediction[0] == 1 else 'no'}")
print(f"probability of buying: {probability[0][1]*100:.2f}%")
```
