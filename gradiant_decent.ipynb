```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y = make_regression(n_samples=4,n_features=1, n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
# Let apply the OLS.
#Ordinary Least Squares (OLS) is a regression technique that finds the line of best fit by minimizing the sum of the squared vertical distances (residuals) between the observed data points and the line.
from sklearn.linear_model import LinearRegression

```

```python
reg = LinearRegression()
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color = 'red')
```

```python
#now apply the gradiant descent assuming the m = 78.35 constant. and b = 0 veriable.
y_pred = ((78.35 * x) + 0).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color = 'red', label='OLS')
plt.plot(x,y_pred,color = '#00a65a',label = 'b = 0') #predicted line.
plt.legend()
plt.show()
```

```python
m = 78.35
b = 0
loss_slop = -2 * np.sum(y - m*x.ravel() - b)
#take learning reate = 0.1

lr = 0.1

step_size = loss_slop*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred1 = ((78.35 * x) + b).reshape(4)

plt.scatter(x,y)
plt.plot(x,reg.predict(x),color = 'red', label='OLS')
plt.plot(x,y_pred1,color = '#A3E4D7',label = f'{b}')
plt.plot(x,y_pred,color = '#00a65a',label = 'b = 0') #predicted line.
plt.legend()
plt.show()
```

- In first step we move closer to our best fit line and now perform another opreation.

```python
loss_slop = -2 * np.sum(y - m*x.ravel() - b)
loss_slop
```

```python
step_size = loss_slop*lr
step_size
```

```python
b = b-step_size
b
```

```python
y_pred2 = ((78.35 * x) + b).reshape(4)

plt.scatter(x,y)
plt.plot(x,reg.predict(x),color = 'red', label='OLS')
plt.plot(x,y_pred1,color = '#A3E4D7',label = f'{b}')
plt.plot(x,y_pred2,color = 'black',label = f'{b}')
plt.plot(x,y_pred,color = '#00a65a',label = 'b = 0') #predicted line.
plt.legend()
plt.show()
```
