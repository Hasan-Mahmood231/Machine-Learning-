# Normalization :

Normalization is a technique used in machine learning to bring all feature values into the same scale, usually between 0 and 1, so that large numbers like salary (in thousands) don’t dominate smaller ones like age. It ensures fair learning and faster model convergence. Standardization, on the other hand, doesn’t bound values between 0 and 1 — instead, it transforms data so it has a mean of 0 and a standard deviation of 1, centering the data around zero. In short, normalization scales values to a fixed range, while standardization shifts and scales data based on its distribution to make it more balanced around the mean.
